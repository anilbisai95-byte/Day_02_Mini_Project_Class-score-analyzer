```python
print("\n"+"="*45)
print("\n----Class Score Analyzer----")
print("\n--Day 2 - Mini Project--")
print("\n"+"="*45)

# 1.Create Student data
students = ["Anil", "Priya", "Ravi", "Divya", "Kiran", "Sunita"]
scores =[87,92,61,74,95,55]


# 2.Core statistics
total = len(scores)
average = sum(scores)/len(scores)
highest = max(scores)
lowest = min(scores)
top_score_index = scores.index(max(scores))
lowest_score_index = scores.index(min(scores))


# 3.Grade classification using comprehension
grades = ["A" if s>=85 else "B" if s>=70 else "C" if s>=55 else "F" for s in scores]


# 4.pass/fail counts
passed = len([s for s in scores if s>=60])
failed = len(scores) -passed

# 5.Normalized to 0-1 range(ML preprocessing)
Normalized = [round(s/highest, 2) for s in scores]


# 6.Print the report
for name, score, grade in zip(students, scores, grades):
    status = "✓ Pass" if score >= 60 else "✗ Fail"
    print(f"  {name:8} | Score:{score:>4} | Grade: {grade} | {status}")
print("-"*45)

print(f" Average : {average:.2f}")
print(f" Highest : {highest} ({students[top_score_index]})")
print(f" Lowest : {lowest} ({students[lowest_score_index]})")
print(f" Passed : {passed}/{total}")
print(f" Failed : {failed}/{total}")
print("-"*45)
print(f" Normalized scores 0-1 scale :")
print(f" {Normalized}")
print("-"*45)
```
